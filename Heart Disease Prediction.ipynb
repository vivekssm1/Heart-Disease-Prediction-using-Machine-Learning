```python
# ============================================================
# BLOCK 1: INSTALL AND IMPORT REQUIRED LIBRARIES
# ============================================================

!pip install kagglehub -q

import os
import kagglehub
import pandas as pd
import numpy as np

# Visualization libraries
import matplotlib.pyplot as plt
import seaborn as sns

# Machine Learning libraries
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

# Evaluation metrics
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

print("All required libraries imported successfully.")
```

```python
# ============================================================
# BLOCK 2: DOWNLOAD DATASET FROM KAGGLE USING KAGGLEHUB
# ============================================================

# Download latest version of the dataset
path = kagglehub.dataset_download(
    "cherngs/heart-disease-cleveland-uci"
)

print("Path to dataset files:", path)
```

```python
# ============================================================
# BLOCK 3: CHECK FILES AVAILABLE IN DOWNLOADED DATASET
# ============================================================

files = os.listdir(path)

print("Files available in dataset:")
for file in files:
    print(file)
```

```python
# ============================================================
# BLOCK 4: LOAD HEART DISEASE DATASET
# ============================================================

file_path = os.path.join(path, "heart_cleveland_upload.csv")

df = pd.read_csv(file_path)

print("Dataset loaded successfully.")

df.head(50)
```

```python
# ============================================================
# BLOCK 5: CHECK DATASET DIMENSIONS
# ============================================================

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("\nDataset Shape:", df.shape)
```

```python
# ============================================================
# BLOCK 6: DISPLAY COLUMN NAMES
# ============================================================

print("Columns available in the dataset:\n")

for column in df.columns:
    print(column)
```

```python
# ============================================================
# BLOCK 7: CHECK DATA TYPES
# ============================================================

print(df.dtypes)
```

```python
# ============================================================
# BLOCK 8: DISPLAY DATASET INFORMATION
# ============================================================

df.info()
```

```python
# ============================================================
# BLOCK 9: CHECK MISSING VALUES
# ============================================================

missing_values = df.isnull().sum()

print("Missing values in each column:\n")
print(missing_values)

print("\nTotal missing values:", missing_values.sum())
```

```python
# ============================================================
# BLOCK 10: HANDLE MISSING VALUES
# ============================================================

# Fill numerical missing values with median
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].median())

print("Missing values after preprocessing:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
# ============================================================
# BLOCK 11: CHECK DUPLICATE RECORDS
# ============================================================

duplicates = df.duplicated().sum()

print("Number of duplicate records:", duplicates)
```

```python
# ============================================================
# BLOCK 12: REMOVE DUPLICATE RECORDS
# ============================================================

df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
# ============================================================
# BLOCK 13: DESCRIPTIVE STATISTICS
# ============================================================

df.describe()
```

```python
# ============================================================
# BLOCK 21: CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(12, 9))

correlation = df.corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt=".3f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap of Heart Disease Dataset")

plt.show()
```

```python
# ============================================================
# BLOCK 22: SEPARATE FEATURES AND TARGET VARIABLE
# ============================================================

# Input features
X = df.drop("condition", axis=1)

# Target variable
y = df["condition"]

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nInput Features:")
print(X.columns.tolist())
```

```python
# ============================================================
# BLOCK 23: TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Features:", X_train.shape)
print("Testing Features :", X_test.shape)

print("Training Labels  :", y_train.shape)
print("Testing Labels   :", y_test.shape)
```

```python
# ============================================================
# BLOCK 24: FEATURE SCALING USING STANDARDIZATION
# ============================================================

scaler = StandardScaler()

# Learn scaling parameters only from training data
X_train_scaled = scaler.fit_transform(X_train)

# Apply same transformation to testing data
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed successfully.")
```

```python
# ============================================================
# BLOCK 25: DISPLAY SCALED TRAINING DATA
# ============================================================

X_train_scaled_df = pd.DataFrame(
    X_train_scaled,
    columns=X.columns
)

X_train_scaled_df.head()
```

```python
# ============================================================
# BLOCK 26: CREATE LOGISTIC REGRESSION MODEL
# ============================================================

model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

print(model)
```

```python
# ============================================================
# BLOCK 27: TRAIN LOGISTIC REGRESSION MODEL
# ============================================================

model.fit(X_train_scaled, y_train)

print("Model training completed successfully.")
```

```python
# ============================================================
# BLOCK 28: PREDICT HEART DISEASE
# ============================================================

y_pred = model.predict(X_test_scaled)

print("Predicted values:")
print(y_pred)
```

```python
# ============================================================
# BLOCK 29: COMPARE ACTUAL AND PREDICTED VALUES
# ============================================================

comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

comparison.head(15)
```

```python
# ============================================================
# BLOCK 30: CALCULATE MODEL PERFORMANCE
# ============================================================

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("Logistic Regression Performance")
print("--------------------------------")

print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1-Score :", round(f1, 4))
```

```python
# ============================================================
# BLOCK 31: DISPLAY PERFORMANCE IN PERCENTAGE
# ============================================================

print("Model Performance")
print("-----------------")

print(f"Accuracy  : {accuracy * 100:.2f}%")
print(f"Precision : {precision * 100:.2f}%")
print(f"Recall    : {recall * 100:.2f}%")
print(f"F1-Score  : {f1 * 100:.2f}%")
```

```python
# ============================================================
# BLOCK 32: CLASSIFICATION REPORT
# ============================================================

print("Classification Report:\n")

print(classification_report(
    y_test,
    y_pred,
    target_names=["No Heart Disease", "Heart Disease"]
))
```

```python
# ============================================================
# BLOCK 33: CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
# ============================================================
# BLOCK 34: VISUALIZE CONFUSION MATRIX
# ============================================================

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No Disease", "Disease"],
    yticklabels=["No Disease", "Disease"]
)

plt.title("Confusion Matrix - Logistic Regression")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")

plt.show()
```

```python
# ============================================================
# BLOCK 35: VISUALIZE MODEL PERFORMANCE
# ============================================================

metrics = ["Accuracy", "Precision", "Recall", "F1-Score"]

scores = [
    accuracy,
    precision,
    recall,
    f1
]

plt.figure(figsize=(8, 5))

plt.bar(metrics, scores)

plt.ylim(0, 1)

plt.title("Logistic Regression Performance")
plt.xlabel("Evaluation Metric")
plt.ylabel("Score")

plt.show()
```

```python
# ============================================================
# BLOCK 36: DISPLAY MODEL FEATURE COEFFICIENTS
# ============================================================

coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_[0]
})

coefficients = coefficients.sort_values(
    by="Coefficient",
    ascending=False
)

coefficients
```

```python
# ============================================================
# BLOCK 37: VISUALIZE FEATURE COEFFICIENTS
# ============================================================

plt.figure(figsize=(9, 6))

sns.barplot(
    data=coefficients,
    x="Coefficient",
    y="Feature"
)

plt.title("Feature Coefficients - Logistic Regression")
plt.xlabel("Coefficient")
plt.ylabel("Feature")

plt.show()
```

```python
# ============================================================
# BLOCK 38: PREDICT HEART DISEASE FOR A NEW PATIENT
# ============================================================

new_patient = pd.DataFrame([{
    "age": 55,
    "sex": 1,
    "cp": 2,
    "trestbps": 140,
    "chol": 240,
    "fbs": 0,
    "restecg": 1,
    "thalach": 150,
    "exang": 0,
    "oldpeak": 1.5,
    "slope": 1,
    "ca": 0,
    "thal": 2
}])

print("New Patient Data:")
display(new_patient)
```

```python
# ============================================================
# BLOCK 39: SCALE NEW PATIENT DATA AND MAKE PREDICTION
# ============================================================

new_patient_scaled = scaler.transform(new_patient)

prediction = model.predict(new_patient_scaled)

if prediction[0] == 1:
    print("Prediction: Heart Disease")
else:
    print("Prediction: No Heart Disease")
```

```python
# ============================================================
# BLOCK 40: DISPLAY PREDICTION PROBABILITY
# ============================================================

probability = model.predict_proba(new_patient_scaled)

print("Probability of No Heart Disease:",
      round(probability[0][0] * 100, 2), "%")

print("Probability of Heart Disease:",
      round(probability[0][1] * 100, 2), "%")
```
